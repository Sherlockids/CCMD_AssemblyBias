```python
def chi_square(x, cov_a, y, cov_b):
    """
    # calculate \chi square of X and Y
    # Assume X with value [x] and covirance matrix [a]; Y with value [y] and covirance matrix [b]
    """
    cov_total  = cov_a + cov_b
    variances = np.diag(cov_total) 
    zero_element = variances == 0
    zero_elements_indices = np.where(variances == 0)[0]
    x = x[~zero_element]
    y = y[~zero_element]
    ### Remove the ith row and column
    cov_total = np.delete(np.delete(cov_total, zero_elements_indices, axis=0), zero_elements_indices, axis=1)
    ### Normalize the covariance matrix ###########################################
    # stdv = np.sqrt(np.diag(cov_total))  # Square root of the diagonal elements
    # outer_diag = np.outer(stdv, stdv)   # Outer product of the diagonal elements
    # corr = cov_total / outer_diag
    # pinv_corr = np.linalg.pinv(corr, rcond = 1e-5)
    # diff = (x - y) / stdv
    # chi_square = np.dot(diff.T, np.dot(pinv_corr, diff))
    ###############################################################################
    diff = x - y
    pinv_cov_total = np.linalg.pinv(cov_total, rcond = 1e-5)
    chi_square = np.dot(diff.T, np.dot(pinv_cov_total, diff))

    return chi_square
```
